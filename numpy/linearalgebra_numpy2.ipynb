## Vector-Vector Multiplication (.) Dot Product

```python
def vector_vector_multiplication(u,v):
    assert u.shape == v.shape
    n = u.shape[0]
    result = 0.0

    for i in range(n):
        result = result + (u[i] * v[i])

    return result

result1 = vector_vector_multiplication(a_vector, b_vector)
result1
        
```

```python
result2 = np.dot(a_vector, b_vector)
result2
```

```python
result3 = a_vector @ b_vector
result3
```

## Matrix-Vector Multiplication

```python
np.random.seed(34)
c_matrix = np.random.randint(low = 1, high = 30, size= [4,4])
c_matrix
```

```python
def matrix_vector_multiplication(U, v):
    assert U.shape[1] == v.shape[0]
    
    num_rows = U.shape[0]
    
    result = np.zeros(num_rows)
    
    for i in range(num_rows):
        result[i] = vector_vector_multiplication(U[i], v)
    
    return result

result = matrix_vector_multiplication(c_matrix, a_vector)
result
```

```python
result = c_matrix.dot(a_vector)
result
```

## Matrix-Matrix Multiplication

```python
np.random.seed(38)
d_matrix = np.random.randint(low = 1, high = 30, size= [2,4])
d_matrix
```

```python
def matrix_matrix_multiplication(U, V):
    assert U.shape[1] == V.shape[0]
    
    num_rows = U.shape[0]
    num_cols = V.shape[1]
    
    result = np.zeros((num_rows, num_cols))
    
    for i in range(num_cols):
        vi = V[:, i]
        Uvi = matrix_vector_multiplication(U, vi)
        result[:, i] = Uvi
    
    return result

result = matrix_matrix_multiplication(d_matrix, c_matrix)
result
```

```python
d_matrix.dot(c_matrix)
```

```python
d_matrix.dot(np.eye(4
                   ))
```
