## Linear Algebra

```python
import numpy as np
```

```python
np.random.seed(35)
a_vector = np.random.randint(low = 0, high = 10, size = 4)
a_vector
```

```python
np.random.seed(43)
b_vector = np.random.randint(low = 0, high = 10, size = 4)
b_vector
```

### Vector Operations

```python
2 * a_vector
```

```python
a_vector + b_vector
```

## Vector-Vector Multiplication 1. (x) Cross Product

```python
a_vector * b_vector
```

## Vector-Vector Multiplication (.) Dot Product

```python
def vector_dot_product(u,v):
    assert u.shape == v.shape
    n = u.shape[0]
    result = 0.0

    for i in range(n):
        result = result + (u[i] * v[i])

    return result

result1 = vector_dot_product(a_vector, b_vector)
result1
        
```

```python
result2 = np.dot(a_vector, b_vector)
result2
```

```python
result3 = a_vector @ b_vector
result3
```
